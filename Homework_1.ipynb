```python
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
import scipy
from scipy import stats

#this sets the size of the plot to something useful
plt.rcParams["figure.figsize"] = (15,10)
```

Simple Statistics

```python
d = stats.norm.rvs(loc = 5.0, scale = .01, size = 1000000)
```

```python
fig, ax = plt.subplots(1, 1)
ax.hist(d,50, density=True)
plt.tick_params(labelsize = 24)
plt.xlim([4.95,5.05])
x = np.linspace(4.95,5.05,1000)
ax.plot(x,stats.norm.pdf(x,loc = 5., scale = 0.01),linewidth = 8,alpha = 0.7)
plt.show()
```

```python
ax = plt.hist(d,50, density=True)
plt.yscale('log')
plt.tick_params(labelsize = 24)
plt.xlim([4.95,5.05])
plt.show()
```

```python
fig, ax = plt.subplots(1, 1)
ax.hist(d,50, density=True)
plt.yscale('log')
plt.tick_params(labelsize = 24)
plt.xlim([4.95,5.05])
x = np.linspace(4.95,5.05,1000)
ax.plot(x,stats.norm.pdf(x,loc = 5., scale = 0.01),linewidth = 8,alpha = 0.7)
plt.show()
```

```python
# Creating a new distribution where the center is 10 points from 0 and where the width is much wider at 1 point.
# Adjusting the generation of random numbers to 100k instead of a million for our normal distribution.

d_new = stats.norm.rvs(loc = 10.0, scale = 1.0, size = 100000) # loc represents the mean of the curve while scale represents the width of the curve
```

```python
fig, ax = plt.subplots(1, 1) # Preparing the plotting area

ax.hist(d_new, 50, density=True) # Generating the histogram for d_new where it'll spit 50 vertical bars. 
# Density set to true so that histogram becomes a probability chart instead of raw counts.

plt.tick_params(labelsize = 24) # Adjusting sizing to 24 so that number on the X and Y axes are large and readable

plt.xlim([6,14]) # Plotting my x range so that window zooms in to focus directly between 6 and 14 on the x axis
# Center is defined at 10, so zooming window in between 6 and 14 allows for viewing the entire normal distribution

x = np.linspace(6, 14, 1000) # Generating sequence of 1000 numbers spaced apart from each other starting from 6 all the way to 14 on the x axis.

ax.plot(x, stats.norm.pdf(x, loc = 10.0, scale = 1.0), linewidth = 8, alpha = 0.7) 
# Calculating theoritcal curve to fit over our normal distribution, while line width adjusts thickness of this curve and alpha allows for it to be see-through.

plt.show() # Python renders the final composition onto screen
```

With a scale of 1.0, the width of the curve goes up to 3 points from the defined center at x = 10. This is due to the normal distribution flattening out and going through 3 standard deviations. 

```python
# Plotting our histogram for d_new again but changing the y axis so that it is in log scale. Haven't added theoretical curve yet. 
# Number of vertical bars, label size, and x axis range stays the same as in the code above. 
ax = plt.hist(d_new, 50, density=True) 
plt.yscale('log')
plt.tick_params(labelsize = 24)
plt.xlim([6, 14])
plt.show()
```

```python
# Now combining the two blocks of code above to add a theoretical curve to our normal distribution for d_new. 
# Includes the change to log scale for the y axis. 

fig, ax = plt.subplots(1, 1)
ax.hist(d_new ,50, density=True)
plt.yscale('log')
plt.tick_params(labelsize = 24)
plt.xlim([6, 14])
x = np.linspace(6, 14, 1000)
ax.plot(x,stats.norm.pdf(x,loc = 10.0, scale = 1.0),linewidth = 8,alpha = 0.7)
plt.show()
```
